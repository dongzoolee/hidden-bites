# 네이버 지도 리뷰 text 확인

장소명과 리뷰 원문을 가볍게 확인하기 위한 notebook입니다.

```python
from pathlib import Path
import json
from html import escape
from IPython.display import HTML, display

dataset_path = Path("../datasets/naver-map-reviews-2026-05-13.json")
if not dataset_path.exists():
    dataset_path = Path("datasets/naver-map-reviews-2026-05-13.json")

dataset = json.loads(dataset_path.read_text(encoding="utf-8"))
locations = dataset["locations"]
rows = [
    {
        "location_name": location["location"]["name"],
        "location_type": location["location"]["type"],
        "category": location["location"].get("category"),
        "rank": review["rank"],
        "visited_date": review.get("visited_date"),
        "text": review["text"],
    }
    for location in locations
    for review in location["reviews"]
]

{
    "dataset_path": str(dataset_path),
    "locations_count": len(locations),
    "reviews_count": len(rows),
}
```

## 장소명 확인

```python
location_rows = [
    {
        "position": item["position"],
        "location_name": item["location"]["name"],
        "location_type": item["location"]["type"],
        "category": item["location"].get("category"),
        "reviews": len(item["reviews"]),
    }
    for item in locations
]

header = "".join(f"<th>{escape(key)}</th>" for key in location_rows[0].keys())
body = "".join(
    "<tr>" + "".join(f"<td>{escape(str(value))}</td>" for value in row.values()) + "</tr>"
    for row in location_rows
)
display(HTML(f"<table><thead><tr>{header}</tr></thead><tbody>{body}</tbody></table>"))
```

## 리뷰 text 확인

```python
columns = ["location_name", "rank", "visited_date", "text"]
header = "".join(f"<th>{escape(column)}</th>" for column in columns)
body = "".join(
    "<tr>" + "".join(f"<td>{escape(str(row[column]))}</td>" for column in columns) + "</tr>"
    for row in rows
)
display(HTML(f"""
<style>
table {{ border-collapse: collapse; width: 100%; font-size: 13px; }}
th, td {{ border: 1px solid #ddd; padding: 8px; vertical-align: top; }}
th {{ background: #f6f6f6; text-align: left; }}
td:nth-child(1) {{ white-space: nowrap; }}
td:nth-child(4) {{ min-width: 420px; line-height: 1.5; }}
</style>
<table><thead><tr>{header}</tr></thead><tbody>{body}</tbody></table>
"""))
```
